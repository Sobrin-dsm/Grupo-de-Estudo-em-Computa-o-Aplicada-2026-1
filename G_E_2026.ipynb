# Entendendo o contexto

Para melhor entender o contexto do banco de dados estudado, foi elaborada uma pesquisa acerca do tema em questão, a qual gerou os seguintes frutos (resumidos):

O contexto é acerca da Citologia Aspirativa (PAAF): Onde um médico inseriu uma agulha fina em uma massa (caroço) na mama, extraiu células e as colocou sob um microscópio. O que há no DataFrame são medições geométricas dessas células.

As 30 colunas são divididas em 10 características básicas, medidas de três formas diferentes:

# 1. As 10 Características Básicas
### Cada uma dessas métricas descreve a aparência física do núcleo das células:

Tamanho: Raio, Perímetro e Área.

Formato: Suavidade, Concavidade, Pontos Côncavos e Simetria.

Complexidade: Textura, Compacidade e Dimensão Fractal (detalhes da borda).

# 2. Como os dados são organizados?
Cada uma das 10 métricas acima aparece em 3 versões:

Média (Mean): A média dessas medidas para todas as células encontradas na amostra do paciente.

Erro (Error): O quanto essas medidas variam entre as células. Se o erro é alto, as células são muito diferentes entre si (o que pode ser um sinal de alerta).

Pior (Worst): A média das 3 maiores medidas encontradas. Na medicina, muitas vezes o "pior caso" é mais importante do que a média geral para detectar um tumor.

# 3. O Objetivo Final
Diferenciar células Benignas (pequenas, lisas e uniformes) de células Malignas (grandes, deformadas, com bordas irregulares e muita variação entre elas).

```python
# 1. Verificar a quantidade de colunas antes da limpeza
colunas_antes = df.shape[1]
print({colunas_antes})

df = df.replace(r'^\s*$', np.nan, regex=True)

colunas_vazias = df.columns[df.isna().all()]
print(f"\nColunas totalmente vazias detectadas ({len(colunas_vazias)}):")
print(list(colunas_vazias))

```

```python
colunas_idade = [c for c in df.columns if 'idade' in c.lower()]
for col in colunas_idade:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    print(f"Tipo NUMÉRICO aplicado na coluna: {col}")
```

```python
radicais_alvo = ['desfecho', 'doenç', 'reaç', 'medicamento']

print("\nSubstituindo nulos por 'Não Informado':")
for radical in radicais_alvo:
    colunas_encontradas = [c for c in df.columns if radical in c.lower()]
    for col in colunas_encontradas:
        # Substitui os NaNs por 'Não Informado'
        df[col] = df[col].fillna('Não Informado')
        # Caso tenha sobrado alguma string vazia ou com espaços, padroniza também
        df[col] = df[col].replace(r'^\s*$', 'Não Informado', regex=True)
    if colunas_encontradas:
        print(f" -> Tratadas as colunas: {colunas_encontradas}")
```

```python
# 1. ENGENHARIA DE DADOS: DIVIDINDO A FAIXA ETÁRIA (One-Hot Encoding)
# ==========================================
# Criando uma coluna binária (0 ou 1) para cada grupo de forma isolada
df['Is_Crianca'] = df['Idade'].apply(lambda x: 1 if x <= 11 else 0)
df['Is_Adolescente'] = df['Idade'].apply(lambda x: 1 if (x > 11 and x <= 17) else 0)
df['Is_Adulto'] = df['Idade'].apply(lambda x: 1 if (x > 17 and x <= 59) else 0)
df['Is_Idoso'] = df['Idade'].apply(lambda x: 1 if x > 59 else 0)


# ==========================================
# 2. OUTRAS VARIÁVEIS NUMÉRICAS (Mantendo o que funcionou)
# ==========================================
colunas_vacinas = [c for c in df.columns if 'vacina' in c.lower() and c != 'Imunobiológico (vacina)']
df['Qtd_Vacinas_Registradas'] = df[colunas_vacinas].apply(lambda row: sum((row != 'Não Informado') & (row.notna())), axis=1)

colunas_reacoes = [c for c in df.columns if 'reaç' in c.lower() or 'evento' in c.lower()]
df['Qtd_Reacoes_Registradas'] = df[colunas_reacoes].apply(lambda row: sum((row != 'Não Informado') & (row.notna())), axis=1)

df['Is_Feminino'] = df['Sexo'].apply(lambda x: 1 if str(x).strip().lower() == 'feminino' else 0)

col_gravidade = [c for c in df.columns if 'gravidade' in c.lower()][0]
df['Is_Grave'] = df[col_gravidade].apply(lambda x: 1 if 'grave' in str(x).lower() and 'não' not in str(x).lower() else 0)


# ==========================================
# 3. NOVA MATRIZ DE CORRELAÇÃO E HEATMAP
# ==========================================
# Substituímos 'Faixa_Etaria_Cod' pelas 4 novas colunas independentes
colunas_para_correlacao = [
    'Idade',
    'Is_Crianca',
    'Is_Adolescente',
    'Is_Adulto',
    'Is_Idoso',
    'Qtd_Vacinas_Registradas',
    'Qtd_Reacoes_Registradas',
    'Is_Feminino'
]

df_numeric = df[colunas_para_correlacao].dropna()

# Calcular a Matriz
matriz_corr = df_numeric.corr()

mascara = np.triu(np.ones_like(matriz_corr, dtype=bool))

# Gerar o Heatmap ampliado
plt.figure(figsize=(12, 10))
sns.heatmap(matriz_corr, annot=True, cmap='coolwarm', fmt=".2f", mask=mascara, linewidths=0.5, vmin=-1, vmax=1)
plt.title('Matriz de Correlação Expandida por Faixa Etária (ESAVI)', fontsize=14, pad=15)
plt.tight_layout()
plt.show()


# ==========================================
# 4. IDENTIFICAÇÃO AUTOMÁTICA DOS NOVOS PARES
# ==========================================
corr_pairs = matriz_corr.unstack().reset_index()
corr_pairs.columns = ['Variavel_1', 'Variavel_2', 'Correlacao']
corr_pairs = corr_pairs[corr_pairs['Variavel_1'] < corr_pairs['Variavel_2']]
corr_pairs['Abs_Correlacao'] = corr_pairs['Correlacao'].abs()
```

```python
variaveis_idade = ['Idade', 'Is_Crianca', 'Is_Adolescente', 'Is_Adulto', 'Is_Idoso']
variaveis_desfecho = ['Qtd_Vacinas_Registradas', 'Qtd_Reacoes_Registradas', 'Is_Feminino']

# 2. Calcular a matriz completa, mas filtrar apenas o cruzamento que nos interessa
# Isso isola o impacto de cada grupo diretamente no desfecho
matriz_focada = df[variaveis_idade + variaveis_desfecho].corr()
matriz_focada = matriz_focada.loc[variaveis_idade, variaveis_desfecho]

# 3. Plotar o Heatmap Retangular (Sem necessidade de máscara triangular aqui)
plt.figure(figsize=(7, 5))
sns.heatmap(
    matriz_focada,
    annot=True,
    cmap='coolwarm',
    fmt=".2f",
    vmin=-1, vmax=1,
    linewidths=0.5
)

plt.title('Impacto das Faixas Etárias nos Desfechos de ESAVI', fontsize=12, pad=15)
plt.tight_layout()
plt.show()
```

Realizada a Exposição de análises de dados multivariados de diversas formas diferentes, incluindo a scatter plot, vista anteriormente

```python
# Selecionar as variáveis numéricas chave e a categórica para a cor
colunas_pairplot = ['Idade', 'Qtd_Vacinas_Registradas', 'Qtd_Reacoes_Registradas', 'Sexo']

# Gerar a Scatter Matrix profissional
sns.set_theme(style="ticks")
g = sns.pairplot(df[colunas_pairplot], hue='Sexo', palette='Set1', diag_kind='kde', plot_kws={'alpha': 0.6})

g.fig.suptitle('Scatter Matrix Multivariada: Idade, Vacinas e Reações por Sexo', y=1.02, fontsize=14, weight='bold')
plt.xticks(rotation=45, ha='right')
plt.show()
```

A Scatter Matrix (Pairplot) analisa simultaneamente as relações bivariadas e as distribuições marginais de Idade, Vacinas e Reações, segmentadas por Sexo. Pela diagonal (curvas de densidade KDE), observa-se que a Idade possui uma distribuição multimodal para ambos os sexos, enquanto a Quantidade de Reações apresenta uma forte assimetria positiva, concentrando-se fortemente entre 2 e 4 notificações. O cruzamento de Idade vs. Vacinas mostra sutilmente que o acúmulo de doses se concentra em idades mais avançadas, sem distinção visual clara entre homens e mulheres nas dispersões.

```python
#Scatter Plot 4D

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Idade',
    y='Qtd_Reacoes_Registradas',
    hue='Sexo',
    style='Is_Grave',
    markers={0: 'o', 1: 'X'}, # 'o' para casos normais, 'X' para casos graves
    s=100,
    alpha=0.7,
    palette='Dark2'
)

plt.title('Relação Multivariada: Idade vs Reações Clínicas (Por Sexo e Gravidade)', fontsize=13, weight='bold', pad=15)
plt.xlabel('Idade do Paciente')
plt.ylabel('Quantidade de Reações Registadas')
plt.legend(title='Fatores Categóricos', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

Este gráfico de dispersão mapeia de forma multivariada a Idade (X) contra a Quantidade de Reações (Y), usando cores para o Sexo. Um ponto analítico crucial nesta amostra é que os casos graves (Is_Grave) não pontuaram no gráfico (exibindo apenas o marcador padrão '0' na legenda), o que indica uma predominância de eventos leves ou moderados. Visualmente, confirma-se o comportamento de patamares horizontais de reações (linhas retas em 2, 4, 6, 8, 10 e 12), demonstrando uma forte tendência de notificações em números pares, com uma presença ligeiramente mais dispersa e elevada de pacientes do sexo feminino.

```python
def agrupar_faixa_etaria(idade):
    if pd.isna(idade): return 'Não Informado'
    elif idade <= 11: return 'Criança'
    elif idade <= 17: return 'Adolescente'
    elif idade <= 59: return 'Adulto'
    else: return 'Idoso'

df['Faixa_Etaria'] = df['Idade'].apply(agrupar_faixa_etaria)

# Configurar o FacetGrid para dividir em colunas baseadas no Atendimento Médico
g = sns.FacetGrid(df, col="AtendimentoMedico", hue="Sexo", palette="muted", height=5, aspect=1.2)

# Mapear o Boxplot dentro de cada faceta
g.map(sns.boxplot, "Faixa_Etaria", "Qtd_Reacoes_Registradas", order=['Criança', 'Adolescente', 'Adulto', 'Idoso'])

g.add_legend(title="Sexo")
g.set_titles("Atendimento Médico: {col_name}")
g.set_axis_labels("Faixa Etária", "Qtd de Reações Registadas")

g.fig.suptitle('Distribuição Multivariada de Reações por Faixa Etária, Sexo e Necessidade Médica', y=1.05, fontsize=14, weight='bold')
plt.show()
```

A análise por facetas revela o maior contraste clínico do estudo. No painel da esquerda (Sem Atendimento Médico), a imensa maioria dos pacientes estabiliza-se na linha de base de 2 reações, independentemente da faixa etária. Já no painel da direita (Com Atendimento Médico), há uma elevação drástica e nítida em todas as caixas (medianas e quartis superiores muito mais altos), provando visualmente que a busca por socorro médico está diretamente atrelada ao volume de reações manifestadas. Nota-se ainda que a categoria 'Adolescente' que buscou atendimento médico é composta exclusivamente por indivíduos do sexo feminino na amostra.
